```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the CSV
df = pd.read_csv("scores.csv")

# Convert date column to datetime
df["date"] = pd.to_datetime(df["date"])

# Plot
plt.figure(figsize=(8,5))
plt.plot(df["date"], df["score_seconds"], marker="o")

plt.title("Score Over Time")
plt.xlabel("Date")
plt.ylabel("Score (seconds)")
plt.grid(True)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load CSV
df = pd.read_csv("scores.csv")

# Convert date column
df["date"] = pd.to_datetime(df["date"])

# Sort by score (ascending = fastest time)
df_sorted = df.sort_values("score_seconds").head(5)

# Plot scoreboard
plt.figure(figsize=(8,5))
plt.barh(df_sorted["name"], df_sorted["score_seconds"], color="#4C72B0")
plt.gca().invert_yaxis()  # Highest rank at top

plt.title("Top 5 Fastest Scores")
plt.xlabel("Score (seconds)")
plt.grid(axis="x", linestyle="--", alpha=0.5)

plt.show()
```
